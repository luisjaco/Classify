# CDX Toolkit
The `CDX Toolkit` provides us with very useful features which we can utilize during development. It 
allows us to easily query the CommonCrawl database and indices.

```python
# Use the following command to get an inaccurate estimate of the result size of a given query.
#   --crawl 2                       indicates that we want to search the 2 most recent crawls.
#   size                            indicates we want the size estimate.
#   "job-boards.greenhouse.io/*"    the url query. "*" is a wildcard character.

!cdxt --crawl 2 size "job-boards.greenhouse.io/*"
```

```python
# Use the following command to get results of a given query.
#   -vv                             leaves debug output.
#   --limit 100                     restricts the output to 100.
#   --crawl 1                       uses the most recent crawl.
#   iter                            mode to search through the database.
#   --csv                           sets to csv output.
#   "job-boards.greenhouse.io/*"    the url query. "*" is a wildcard character.

!cdxt --limit 10 --crawl 2 iter --csv "job-boards.greenhouse.io/*" 
```
